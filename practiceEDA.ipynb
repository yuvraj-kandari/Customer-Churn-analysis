```python
#converted 0 and 1 value to senior citizen yes or no to make it easy to understand

def conv(value):
    if value==1:
        return "yes"
    else :
        return "no"

df['SeniorCitizen'] = df['SeniorCitizen'].apply(conv)       
```

```python
#group by value 
gd= df.groupby("Churn").agg({"Churn":"count"})
gd
```

```python
#pie chart with labels and percentage and we can see that 26 percentage of people have churned out
plt.figure(figsize=(3,4))
plt.pie(gd['Churn'], labels=gd.index ,autopct="%1.2f%%")
plt.title("Perecentage of Churned Customers",fontsize =10)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create percentage table
churn_pct = pd.crosstab(
    df["SeniorCitizen"],
    df["Churn"],
    normalize="index"
) * 100

# Plot stacked bar
ax = churn_pct.plot(
    kind="bar",
    stacked=True,
    figsize=(5, 4)
)

# Add percentage labels
for container in ax.containers:
    labels = [f"{bar.get_height():.1f}%" if bar.get_height() > 0 else ""
              for bar in container]
    ax.bar_label(container, labels=labels, label_type="center")

plt.title("Churn Percentage by Senior Citizen Status")
plt.xlabel("Senior Citizen")
plt.ylabel("Percentage")
plt.xticks([0, 1], ["Non-Senior Citizen", "Senior Citizen"], rotation=0)
plt.legend(title="Churn")
plt.ylim(0, 100)

plt.show()
```
